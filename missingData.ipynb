```python
import pandas as pd
import numpy as np
```

# finding missing data

```python
data = {
    'A': [1,2,np.nan,4,5],
    'B': [1,2,3,4,5],
    'C': [1,2,3,np.nan,np.nan],
    'D': [1,np.nan,np.nan,np.nan,5]
}

df = pd.DataFrame(data)
df
```

```python
df.isna()
```

```python
df.isna().sum()
```

```python
df.isna().any()
```

# remove missing data

```python
df
```

```python
df.dropna()
```

```python
# no rows null value they use it
df.dropna(thresh=3)
```

# filling missing data

```python
df.fillna(0)
```

```python
# har column ke liye diffrent column

values = {'A':0,'B':100,"C":400,"D":500}
df.fillna(value=values)
```
